# 02 — Feature policy and frozen split

Inspect the leakage-safe feature set and create the shared stratified train/test manifest. This deterministic split is reused by every model.

```python
from jev_risk.cli import run_split
from jev_risk.config import DEFAULT_DATA_PATH, MODEL_FEATURES, RESULTS_DIR

split = run_split(DEFAULT_DATA_PATH, RESULTS_DIR / 'splits' / 'split_manifest.csv')
print('Features:', MODEL_FEATURES)
print(f'Train rows: {len(split.train_ids):,}; test rows: {len(split.test_ids):,}')
print(f'Train positive rate: {split.y_train.mean():.4f}; test positive rate: {split.y_test.mean():.4f}')
print('Dataset SHA-256:', split.dataset_sha256)
```
